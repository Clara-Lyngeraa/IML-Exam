## Data
The following cell constructs and shows the data for the exercise. The data simulates growth (in meters) of one of the fastest growing trees, the Thuja Green Giant, each year. Scientists have observed and reported the growth of the tree for 7 years (`X_train`
 and `y_train`
), and now want to predict the future growth. 
The objective is to assist in making predictions based on this data. Additional data from another group has been provided to validate the hypothesis.
The scientists assume a polynomial relationship. 


```python
### TASK 5: DATA CHANGED TO 2ND ORDER POLYNOMIAL!!!
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
# Set a random seed for reproducibility
np.random.seed(42)

# Generate synthetic data
n_samples = 100
X = np.linspace(0, 10, n_samples).reshape(-1, 1)
y_true = X.ravel()**2 + 1.5 * X.ravel() - 3
noise = np.random.normal(0, 1, n_samples)
y = y_true + noise

# Split the data into training and test sets
split_index = int(0.7 * n_samples)
X_train, X_test = X[:split_index], X[split_index:]
y_train, y_test = y[:split_index], y[split_index:]


# Plot the results
plt.figure(figsize=(10, 6))
plt.scatter(X_train, y_train, color='blue', label='Training data')
plt.scatter(X_test, y_test, color='red', label='Test data')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
```

```python
def get_design_matrix(X, degree):
    return PolynomialFeatures(degree).fit_transform(X)

def predict(X, weights, degree):
    design_matrix = get_design_matrix(X, degree)
    return design_matrix @ weights

def polynomial_regression(X_train, X_test, y_train, degree):
    """
    Create and train a model of desired order and use it to predict the growth of the trees.

    :param X: Vector of combined observed years).
    :param y: Vector of combined observed height.
    :param degree: Degree of the model.
    
    :return: Vector containing prediction for training data, vector containing prediction for test data.
    """
    design_matrix = get_design_matrix(X_train, degree)
    # print("X_train: \n", X_train)
    # print("design_matrix: \n", design_matrix)

    weights, _, _, _ = np.linalg.lstsq(design_matrix, y_train)

    y_train_predicted = predict(X_train, weights, degree)
    y_test_predicted = predict(X_test, weights, degree)

    return y_train_predicted, y_test_predicted
```

```python
# Visualization of polynomial regression with hardcoded order = 2
# Combined train and test data, red line shows prediction for train and test data
plt.scatter(X_train, y_train, color='gray')
plt.scatter(X_test, y_test, color='black')
predicted_y_train, predicted_y_test = polynomial_regression(X_train, X_test, y_train, 2)
plt.plot(X_train, predicted_y_train, color='red')
plt.plot(X_test, predicted_y_test, color='red')
plt.show()
```

```python
# Visualization of only the test data prediction
plt.scatter(X_test, y_test, color='blue')
plt.plot(X_test, predicted_y_test, color='red')
plt.show()
```

```python
def compute_mse(y_true, y_pred):
    """Compute Mean Squared Error between true and predicted values."""
    return np.mean((y_true-y_pred)**2)

# Train and evaluate linear models with different polynomial features

degrees = [1, 2, 3, 4, 5, 6]
train_pred = []
test_pred = []
train_error = []
test_error = []

#write code/solution here ...
for degree in range(len(degrees)): 
    pred_y_train, pred_y_test = polynomial_regression(X_train, X_test, y_train, degrees[degree])

    train_pred.append(pred_y_train)
    test_pred.append(pred_y_test)
    
    mse_train = compute_mse(y_train, pred_y_train)
    mse_test = compute_mse(y_test, pred_y_test)

    train_error.append(mse_train)
    test_error.append(mse_test)


```
